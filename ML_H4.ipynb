```python
#Q4
import numpy as np
from sklearn.decomposition import PCA

# Step 1: Define the dataset
data = np.array([
    [0, 2],
    [2, 4],
    [4, 4],
    [5, 4],
    [7, 6]
])

# Step 2: Perform PCA
pca = PCA(n_components=2)
pca.fit(data)

# Step 3: Get principal component directions (weight vectors)
components = pca.components_

# Display the results
print("Principal Component Directions (Weight Vectors):")
print(f"PC1: {components[0]}")
print(f"PC2: {components[1]}")
```
